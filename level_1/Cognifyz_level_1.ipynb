```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Cognifyz Data Science Intership")
print("Level 1 - starting Data Analysis")
```

```python

import pandas as pd

from google.colab import files
uploaded = files.upload()

filename = next(iter(uploaded))
df = pd.read_csv(filename)

print("Dataset loaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

df.head()
```

```python

# Task 1: Explore the dataset structure

print("Dataset shape (rows, columns):")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())
```

```python

# Check missing values in each column

print("Missing values in each column:")
print(df.isnull().sum())

print("\nTotal missing values:")
print(df.isnull().sum().sum())
```

```python

# Fill missing cuisine values

df["Cuisines"] = df["Cuisines"].fillna("Unknown")

print("Missing values after cleaning:")
print(df.isnull().sum())

print("\nMissing cuisine values remaining:")
print(df["Cuisines"].isnull().sum())
```

```python

# Inspect data types

print(df.dtypes)
```

```python

# Convert Aggregate rating to numeric

df["Aggregate rating"] = pd.to_numeric(
    df["Aggregate rating"], errors="coerce"
)

print("Aggregate rating data type:")
print(df["Aggregate rating"].dtype)

print("\nRating summary:")
print(df["Aggregate rating"].describe())
```

```python

# Count restaurants by rating text

print("Rating text distribution:")
print(df["Rating text"].value_counts())

# Plot the distribution of numerical ratings
plt.figure(figsize=(8, 5))
plt.hist(df["Aggregate rating"].dropna(), bins=10, edgecolor="black")
plt.title("Distribution of Restaurant Ratings")
plt.xlabel("Aggregate Rating")
plt.ylabel("Number of Restaurants")
plt.show()
```

```python

# Save the cleaned dataset

df.to_csv("Cognifyz_Level_1_Cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```

```python

# Level 1 - Task 2: Descriptive Analysis

print("Mean restaurant rating:")
print(df["Aggregate rating"].mean())

print("\nMedian restaurant rating:")
print(df["Aggregate rating"].median())

print("\nStandard deviation of restaurant ratings:")
print(df["Aggregate rating"].std())
```

```python

# Find the top 10 cities with the most restaurants

top_cities = df["City"].value_counts().head(10)

print("Top 10 cities with the most restaurants:")
print(top_cities)

# Visualize the results
plt.figure(figsize=(10, 6))
top_cities.plot(kind="bar", edgecolor="black")
plt.title("Top 10 Cities by Number of Restaurants")
plt.xlabel("City")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python

# Find the 10 most common cuisines

cuisine_counts = (
    df["Cuisines"]
    .dropna()
    .str.split(", ")
    .explode()
    .value_counts()
    .head(10)
)

print("Top 10 most common cuisines:")
print(cuisine_counts)

# Plot the results
plt.figure(figsize=(10, 6))
cuisine_counts.plot(kind="bar", edgecolor="black")
plt.title("Top 10 Most Common Cuisines")
plt.xlabel("Cuisine")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python

# Level 1 - Task 3: Geospatial Analysis

plt.figure(figsize=(10, 6))

plt.scatter(
    df["Longitude"],
    df["Latitude"],
    c=df["Aggregate rating"],
    cmap="viridis",
    alpha=0.5
)

plt.colorbar(label="Aggregate Rating")
plt.title("Geographical Distribution of Restaurants")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.tight_layout()
plt.show()
```

```python

# Analyze restaurant distribution and ratings by city

city_analysis = df.groupby("City").agg(
    Restaurant_Count=("Restaurant ID", "count"),
    Average_Rating=("Aggregate rating", "mean")
)

city_analysis = city_analysis.sort_values(
    by="Restaurant_Count", ascending=False
)

print("Top 10 cities by restaurant count:")
display(city_analysis.head(10).round(2))
```

```python

# Correlation between geographic coordinates and ratings

correlations = df[
    ["Longitude", "Latitude", "Aggregate rating"]
].corr()

print("Correlation matrix:")
display(correlations.round(3))
```

```python

# Save the final cleaned Level 1 dataset

df.to_csv("Cognifyz_Level_1_Cleaned.csv", index=False)

print("Level 1 cleaned dataset saved successfully!")
```
