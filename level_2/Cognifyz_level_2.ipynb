```python

import pandas as pd

from google.colab import files
uploaded = files.upload()

filename = next(iter(uploaded))
df = pd.read_csv(filename)

print("Dataset loaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

df.head()
```

```python

# Task 1: Explore the dataset structure

print("Dataset shape (rows, columns):")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())
```

```python

# Check missing values in each column

print("Missing values in each column:")
print(df.isnull().sum())

print("\nTotal missing values:")
print(df.isnull().sum().sum())
```

```python

# Fill missing cuisine values

df["Cuisines"] = df["Cuisines"].fillna("Unknown")

print("Missing values after cleaning:")
print(df.isnull().sum())

print("\nMissing cuisine values remaining:")
print(df["Cuisines"].isnull().sum())
```

```python

# Inspect data types

print(df.dtypes)
```

```python

# Convert Aggregate rating to numeric

df["Aggregate rating"] = pd.to_numeric(
    df["Aggregate rating"], errors="coerce"
)

print("Aggregate rating data type:")
print(df["Aggregate rating"].dtype)

print("\nRating summary:")
print(df["Aggregate rating"].describe())
```

```python

# Count restaurants by rating text

print("Rating text distribution:")
print(df["Rating text"].value_counts())

# Plot the distribution of numerical ratings
plt.figure(figsize=(8, 5))
plt.hist(df["Aggregate rating"].dropna(), bins=10, edgecolor="black")
plt.title("Distribution of Restaurant Ratings")
plt.xlabel("Aggregate Rating")
plt.ylabel("Number of Restaurants")
plt.show()
```

```python

# Save the cleaned dataset

df.to_csv("Cognifyz_Level_1_Cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```

```python

# Level 1 - Task 2: Descriptive Analysis

print("Mean restaurant rating:")
print(df["Aggregate rating"].mean())

print("\nMedian restaurant rating:")
print(df["Aggregate rating"].median())

print("\nStandard deviation of restaurant ratings:")
print(df["Aggregate rating"].std())
```

```python

# Find the top 10 cities with the most restaurants

top_cities = df["City"].value_counts().head(10)

print("Top 10 cities with the most restaurants:")
print(top_cities)

# Visualize the results
plt.figure(figsize=(10, 6))
top_cities.plot(kind="bar", edgecolor="black")
plt.title("Top 10 Cities by Number of Restaurants")
plt.xlabel("City")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python

# Find the 10 most common cuisines

cuisine_counts = (
    df["Cuisines"]
    .dropna()
    .str.split(", ")
    .explode()
    .value_counts()
    .head(10)
)

print("Top 10 most common cuisines:")
print(cuisine_counts)

# Plot the results
plt.figure(figsize=(10, 6))
cuisine_counts.plot(kind="bar", edgecolor="black")
plt.title("Top 10 Most Common Cuisines")
plt.xlabel("Cuisine")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python

# Level 1 - Task 3: Geospatial Analysis

plt.figure(figsize=(10, 6))

plt.scatter(
    df["Longitude"],
    df["Latitude"],
    c=df["Aggregate rating"],
    cmap="viridis",
    alpha=0.5
)

plt.colorbar(label="Aggregate Rating")
plt.title("Geographical Distribution of Restaurants")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.tight_layout()
plt.show()
```

```python

# Analyze restaurant distribution and ratings by city

city_analysis = df.groupby("City").agg(
    Restaurant_Count=("Restaurant ID", "count"),
    Average_Rating=("Aggregate rating", "mean")
)

city_analysis = city_analysis.sort_values(
    by="Restaurant_Count", ascending=False
)

print("Top 10 cities by restaurant count:")
display(city_analysis.head(10).round(2))
```

```python

# Correlation between geographic coordinates and ratings

correlations = df[
    ["Longitude", "Latitude", "Aggregate rating"]
].corr()

print("Correlation matrix:")
display(correlations.round(3))
```

```python

# Save the final cleaned Level 1 dataset

df.to_csv("Cognifyz_Level_1_Cleaned.csv", index=False)

print("Level 1 cleaned dataset saved successfully!")
```

```python

# Level 2 - Task 1: Table Booking and Online Delivery

import pandas as pd

df = pd.read_csv("Cognifyz_Level_1_Cleaned.csv")

print("Table Booking Counts:")
print(df["Has Table booking"].value_counts())

print("\nOnline Delivery Counts:")
print(df["Has Online delivery"].value_counts())
```

```python

# Calculate percentages

table_booking = df["Has Table booking"].value_counts(normalize=True) * 100
online_delivery = df["Has Online delivery"].value_counts(normalize=True) * 100

print("Table Booking Percentages:")
print(table_booking.round(2))

print("\nOnline Delivery Percentages:")
print(online_delivery.round(2))
```

```python

import matplotlib.pyplot as plt

# Compare table booking and online delivery
services = ["Table Booking", "Online Delivery"]
yes_percentages = [12.12, 25.66]
no_percentages = [87.88, 74.34]

plt.figure(figsize=(8, 5))
plt.bar(services, yes_percentages, label="Yes")
plt.bar(services, no_percentages, bottom=yes_percentages, label="No")

plt.ylabel("Percentage of Restaurants (%)")
plt.title("Table Booking vs Online Delivery")
plt.ylim(0, 100)
plt.legend()
plt.show()
```

```python

# Compare average ratings by online delivery

delivery_ratings = df.groupby("Has Online delivery")["Aggregate rating"].mean()

print("Average Rating by Online Delivery:")
print(delivery_ratings.round(2))

delivery_ratings.plot(kind="bar", figsize=(7, 5))

plt.title("Average Rating: Online Delivery vs No Delivery")
plt.xlabel("Online Delivery")
plt.ylabel("Average Rating")
plt.xticks(rotation=0)
plt.show()
```

```python

# Compare average ratings with and without table booking

booking_ratings = df.groupby("Has Table booking")["Aggregate rating"].mean()

print("Average Ratings by Table Booking:")
print(booking_ratings.round(2))

booking_ratings.plot(kind="bar", figsize=(7, 5))
plt.title("Average Ratings: Table Booking vs No Table Booking")
plt.xlabel("Has Table Booking")
plt.ylabel("Average Rating")
plt.xticks(rotation=0)
plt.show()
```

```python

# Online delivery availability by price range

delivery_by_price = pd.crosstab(
    df["Price range"],
    df["Has Online delivery"],
    normalize="index"
) * 100

print("Online Delivery Availability by Price Range (%):")
print(delivery_by_price.round(2))

delivery_by_price.plot(kind="bar", figsize=(8, 5))
plt.title("Online Delivery by Price Range")
plt.xlabel("Price Range (1 = Low, 4 = High)")
plt.ylabel("Percentage of Restaurants (%)")
plt.xticks(rotation=0)
plt.legend(title="Online Delivery")
plt.show()
```

```python

# Level 2 - Task 2: Price Range Analysis

# Count restaurants in each price range
price_counts = df["Price range"].value_counts().sort_index()

print("Number of Restaurants in Each Price Range:")
print(price_counts)

print("\nMost Common Price Range:")
print(price_counts.idxmax())

# Plot the counts
price_counts.plot(kind="bar", figsize=(7, 5))
plt.title("Most Common Price Range")
plt.xlabel("Price Range (1 = Low, 4 = High)")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=0)
plt.show()
```

```python

# Calculate average rating for each price range

avg_rating_by_price = df.groupby("Price range")["Aggregate rating"].mean()

print("Average Rating for Each Price Range:")
print(avg_rating_by_price.round(2))

# Plot average ratings
avg_rating_by_price.plot(kind="bar", figsize=(7, 5))

plt.title("Average Restaurant Rating by Price Range")
plt.xlabel("Price Range (1 = Low, 4 = High)")
plt.ylabel("Average Rating")
plt.xticks(rotation=0)
plt.show()
```

```python


# Find the color associated with the highest average rating

highest_rating_price = avg_rating_by_price.idxmax()

print("Price range with the highest average rating:", highest_rating_price)

highest_color = df.groupby("Price range")["Rating color"].agg(
    lambda x: x.mode().iloc[0]
)

print("Color for the highest-rated price range:",
      highest_color.loc[highest_rating_price])
```

```python

# Level 2 - Task 3: Feature Engineering

# Create restaurant name and address length features
df["Restaurant Name Length"] = df["Restaurant Name"].fillna("").str.len()
df["Address Length"] = df["Address"].fillna("").str.len()

# Encode Yes/No columns as 1/0
df["Has Table Booking"] = df["Has Table booking"].map({"Yes": 1, "No": 0})
df["Has Online Delivery"] = df["Has Online delivery"].map({"Yes": 1, "No": 0})

# Display the new features
print(df[[
    "Restaurant Name",
    "Restaurant Name Length",
    "Address Length",
    "Has Table Booking",
    "Has Online Delivery"
]].head())

print("\nFeature engineering completed!")
```

```python

# Check the new features

new_features = [
    "Restaurant Name Length",
    "Address Length",
    "Has Table Booking",
    "Has Online Delivery"
]

print("Missing values in new features:")
print(df[new_features].isnull().sum())

print("\nFeature summary:")
print(df[new_features].describe().round(2))
```
